### **Question 21**
(Visualization) IMDB_Rating ka distribution dekhne ke liye ek neat Histogram plot karein.

```python
plt.figure(figsize=(9, 5))
sns.histplot(df['IMDB_Rating'], bins=20, kde=True, color='steelblue', edgecolor='white')
plt.title('Distribution of IMDB Ratings', fontsize=14, fontweight='bold')
plt.xlabel('IMDB Rating')
plt.ylabel('Number of Movies')
plt.tight_layout()
plt.show()
```

### **Question 22**
(Visualization) Top 10 Directors aur unki total movies ka ek attractive Bar Chart banayein.

```python
top_directors = df['Director'].value_counts().head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_directors.values, y=top_directors.index, hue=top_directors.index,
            palette='viridis', legend=False)
for i, v in enumerate(top_directors.values):
    plt.text(v + 0.1, i, str(v), va='center')
plt.title('Top 10 Directors by Number of Movies', fontsize=14, fontweight='bold')
plt.xlabel('Number of Movies')
plt.ylabel('Director')
plt.tight_layout()
plt.show()
```

### **Question 24**

### **Question 30**
Dataset ko poora clean karne ke baad safely ek nayi csv file 'cleaned_movies.csv' ke naam se save karein.

```python
df_clean = df.copy()

df_clean = df_clean.drop_duplicates()
df_clean['Certificate'] = df_clean['Certificate'].fillna('Not Rated')
df_clean['Meta_score'] = df_clean['Meta_score'].fillna(df_clean['Meta_score'].median())

print("Missing values left:", df_clean.isnull().sum().sum())
df_clean.to_csv('cleaned_movies.csv', index=False)
print("Saved as cleaned_movies.csv with shape", df_clean.shape)
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
drama_df = df[df['Genre'].str.contains('Drama')].assign(Genre_Group='Drama')
action_df = df[df['Genre'].str.contains('Action')].assign(Genre_Group='Action')
combined = pd.concat([drama_df, action_df])   # a movie with both genres appears in both groups

plt.figure(figsize=(8, 6))
sns.violinplot(data=combined, x='Genre_Group', y='Gross', hue='Genre_Group',
               palette='Set2', inner='box', legend=False)
plt.title('Gross Revenue: Drama vs Action', fontsize=14, fontweight='bold')
plt.xlabel('Genre')
plt.ylabel('Gross Revenue ($)')
plt.tight_layout()
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top5 = df.nlargest(5, 'IMDB_Rating')[['Series_Title', 'IMDB_Rating', 'Meta_score']].copy()
top5['IMDB_x10'] = top5['IMDB_Rating'] * 10

x = np.arange(len(top5))
w = 0.38
plt.figure(figsize=(11, 6))
plt.bar(x - w/2, top5['IMDB_x10'], w, label='IMDB Rating (x10)', color='goldenrod')
plt.bar(x + w/2, top5['Meta_score'], w, label='Meta_score', color='slateblue')
plt.xticks(x, top5['Series_Title'], rotation=20, ha='right')
plt.title('Top 5 Rated Movies: IMDB (scaled to 100) vs Meta_score', fontsize=14, fontweight='bold')
plt.ylabel('Score (out of 100)')
plt.ylim(0, 105)
plt.legend()
plt.tight_layout()
plt.show()
```
